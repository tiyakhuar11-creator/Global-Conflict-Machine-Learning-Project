Fill na is used to fill Nulls And Inplace true is used to changes in original data.

```python
df['Alliance_A'].fillna(Alliance_M,inplace=True)
```

```python
df['Alliance_B'].fillna(A_Mode2,inplace=True)

df['Resource_Dispute'].fillna(Res_M3,inplace=True)
```

```python
df1 = df[['Country_A','Country_B','Conflict_Type','Year','Duration_Days','Military_Deaths_A','Military_Deaths_B','Refugees_Millions','Terrain_Type','Population_A_Millions',
'Population_B_Millions','GDP_A_Billions','GDP_B_Billions','Civilian_Deaths','Economic_Loss_USD_Billions','Resource_Dispute', 'Alliance_A','Alliance_B','Weapons_Used','Air_Strikes','Naval_Battles','Ceasefire','Outcome','UN_Involvement']]
```

labelencoding to covert text data into numerical.

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df1.dtypes
```

we apply for loop to convert all text into numerical.

```python
for i in df1:
    if df1[i].dtype == 'object':
        df1[i] = le.fit_transform( df1[i])
        print(df1)
```

```python
df1.dtypes
```

Feature selection

```python
x=df1.drop('UN_Involvement', axis=1)   # x features
x
```

Target Variable

```python
y = df1['UN_Involvement']      # Target variable
y
```

Train_Test_Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_train
```

Scalling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model .fit(x_train,y_train)   # training model
y_pred= model.predict(x_test)   # predict is a function of prediction. y_pred = Predicted answers generated by the model.
```

```python
from sklearn.metrics import accuracy_score 
```

we check are model predicting y_pred is match the real y_test so the accuracy is 50 neutral

```python
round(accuracy_score(y_pred,y_test),2)*100
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

y_test = Actual (real) answers from the dataset.
y_pred = Predicted answers generated by the model.

We store the predictions in a variable named y_pred.

```python
confusion_matrix(y_pred,y_test)
```

```python
classification_report(y_pred,y_test)
```
